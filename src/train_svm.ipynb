```python
import numpy as np
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
import joblib
```

```python
data = np.load("D:/ErgoSense/dataset/bilstm_embeddings.npz")

X_train = data["X_train"]
y_train = data["y_train"]

X_val = data["X_val"]
y_val = data["y_val"]

X_train.shape, X_val.shape
```

```python
svm_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svc", SVC(kernel="rbf", C=10, gamma="scale", probability=True))
])
```

```python
svm_pipeline.fit(X_train, y_train)
print("SVM training complete!")
```

```python
preds = svm_pipeline.predict(X_val)

print(classification_report(y_val, preds))
print(confusion_matrix(y_val, preds))
```

```python
joblib.dump(svm_pipeline, "D:/ErgoSense/models/svm_pipeline.pkl")
print("Saved to models/svm_pipeline.pkl")
```
